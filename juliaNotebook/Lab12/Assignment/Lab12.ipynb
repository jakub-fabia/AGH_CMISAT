## Zadanie 1

```julia
# Częstotliwość próbkowania
Fs = 1024
# Wektor wszyskich x próbek
t = 0:1/(Fs-1):1
# Wektor próbek
x = sin.(2*pi*t*200) + 2* sin.(2*pi*t*400)
sticks((abs.(fft(x))))
```

Widoczne wysokie amplitudy dla częstotliwości 200 i 400 Hz oraz symetrycznego odbicia tych częstotliwości.
